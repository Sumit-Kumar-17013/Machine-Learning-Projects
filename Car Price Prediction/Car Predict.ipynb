```python
# On hot encoding
df = pd.get_dummies(df , columns=['model' , 'transmission' , 'fuelType'] , drop_first= True)
```

```python
#train - test model
from sklearn.model_selection import train_test_split
x = df.drop(columns= ['price'] , axis = 1)
y = df['price']

x_train , x_test , y_train , y_test = train_test_split(x , y , test_size = 0.2 , random_state = 42 , shuffle = True)
```

```python
#standeerd Scaling
from sklearn.preprocessing import StandardScaler
scale = StandardScaler()
x_train = scale.fit_transform(x_train)
x_test = scale.transform(x_test)
```

```python
#linear Regression
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train , y_train )

L_pred = model.predict(x_test)
print('Predicated Value : ' , L_pred[:5].round(2))

#metrics of this model
from sklearn.metrics import mean_absolute_error , mean_squared_error , r2_score
maeL = mean_absolute_error(y_test , L_pred)
mseL = mean_squared_error(y_test , L_pred)
rmseL = np.sqrt(mseL)
r2L = r2_score(y_test , L_pred)

print('Mean Absolute Error : ' , round(maeL , 2))
print('Mean Squared Error : ' , round(mseL ,2))
print('Root Mean Squared Error : ' , round(rmseL , 2))
print('r2-Score : ' , round(r2L , 2))
```

```python
#another model train by this model
from sklearn.neighbors import KNeighborsRegressor
modelK = KNeighborsRegressor(n_neighbors = 4)
modelK.fit(x_train , y_train)
KN_pred = modelK.predict(x_test)

print('Predicated Value : ' , KN_pred[:5].round(2))

#metrics of this model
from sklearn.metrics import mean_absolute_error , mean_squared_error , r2_score
maeKN = mean_absolute_error(y_test , KN_pred)
mseKN = mean_squared_error(y_test , KN_pred)
rmseKN = np.sqrt(mseKN)
r2KN = r2_score(y_test , KN_pred)

print('Mean Absolute Error : ' , round(maeKN , 2))
print('Mean Squared Error : ' , round(mseKN ,2))
print('Root Mean Squared Error : ' , round(rmseKN , 2))
print('r2-Score : ' , round(r2KN , 2))
```

```python
#another model train by this model
from sklearn.tree import DecisionTreeRegressor
modelT = DecisionTreeRegressor(max_depth= 4)
modelT.fit(x_train , y_train)
T_pred = modelT.predict(x_test)

print('Predicated Value : ' , T_pred[:5].round(2))

#metrics of this model
from sklearn.metrics import mean_absolute_error , mean_squared_error , r2_score
maeT = mean_absolute_error(y_test , T_pred)
mseT = mean_squared_error(y_test , T_pred)
rmseT = np.sqrt(mseT)
r2T = r2_score(y_test , T_pred)

print('Mean Absolute Error : ' , round(maeT , 2))
print('Mean Squared Error : ' , round(mseT ,2))
print('Root Mean Squared Error : ' , round(rmseT , 2))
print('r2-Score : ' , round(r2T , 2))
```

```python
#another model train by this model
from sklearn.ensemble import RandomForestRegressor
modelR = RandomForestRegressor(n_estimators = 100)
modelR.fit(x_train , y_train)
R_pred = modelR.predict(x_test)

print('Predicated Value : ' , R_pred[:5].round(2))

#metrics of this model
from sklearn.metrics import mean_absolute_error , mean_squared_error , r2_score
maeR = mean_absolute_error(y_test , R_pred)
mseR = mean_squared_error(y_test , R_pred)
rmseR = np.sqrt(mseR)
r2R = r2_score(y_test , R_pred)

print('Mean Absolute Error : ' , round(maeR , 2))
print('Mean Squared Error : ' , round(mseR ,2))
print('Root Mean Squared Error : ' , round(rmseR , 2))
print('r2-Score : ' , round(r2R , 2))
```

```python
#another model train by this model
from sklearn.svm import SVR
modelS = SVR(kernel= 'linear')
modelS.fit(x_train , y_train)
S_pred = modelS.predict(x_test)

print('Predicated Value : ' , S_pred[:5].round(2))

#metrics of this model
from sklearn.metrics import mean_absolute_error , mean_squared_error , r2_score
maeS = mean_absolute_error(y_test , S_pred)
mseS = mean_squared_error(y_test , S_pred)
rmseS = np.sqrt(mseS)
r2S = r2_score(y_test , S_pred)

print('Mean Absolute Error : ' , round(maeS , 2))
print('Mean Squared Error : ' , round(mseS ,2))
print('Root Mean Squared Error : ' , round(rmseS , 2))
print('r2-Score : ' , round(r2S , 2))
```

```python
#another model train by this model
from sklearn.linear_model import Ridge
modelRid = Ridge(alpha = 100)
modelRid.fit(x_train , y_train)
RID_pred = modelRid.predict(x_test)

print('Predicated Value : ' , RID_pred[:5].round(2))

#metrics of this model
from sklearn.metrics import mean_absolute_error , mean_squared_error , r2_score
maeRID = mean_absolute_error(y_test , RID_pred)
mseRID = mean_squared_error(y_test , RID_pred)
rmseRID = np.sqrt(mseRID)
r2RID = r2_score(y_test , RID_pred)

print('Mean Absolute Error : ' , round(maeRID , 2))
print('Mean Squared Error : ' , round(mseRID ,2))
print('Root Mean Squared Error : ' , round(rmseRID , 2))
print('r2-Score : ' , round(r2RID , 2))
```

```python
#another model train by this model
from sklearn.linear_model import Lasso
modelLAS = Lasso(alpha = 0.01)
modelLAS.fit(x_train , y_train)
LAS_pred = modelLAS.predict(x_test)

print('Predicated Value : ' , LAS_pred[:5].round(2))

#metrics of this model
from sklearn.metrics import mean_absolute_error , mean_squared_error , r2_score
maeLAS = mean_absolute_error(y_test , LAS_pred)
mseLAS = mean_squared_error(y_test , LAS_pred)
rmseLAS = np.sqrt(mseLAS)
r2LAS = r2_score(y_test , LAS_pred)

print('Mean Absolute Error : ' , round(maeLAS , 2))
print('Mean Squared Error : ' , round(mseLAS ,2))
print('Root Mean Squared Error : ' , round(rmseLAS , 2))
print('r2-Score : ' , round(r2LAS , 2))
```

```python
result = pd.DataFrame({
    'Models' : ["Linear", "KNN", "Decision Tree", "Random Forest", "SVM", "Ridge", "Lasso"],
    'R2-Score' : [r2L , r2KN , r2T , r2R , r2S , r2RID , r2LAS],
    'MAE-Error' : [maeL , maeKN , maeT , maeR , maeS , maeRID , maeLAS],
    'MSE-Error' : [mseL , mseKN , mseT , mseR , mseS , mseRID , mseLAS],
    'RMSE-Error' : [rmseL , rmseKN , rmseT , rmseR , rmseS , rmseRID , rmseLAS],
})
print(result)
```

```python
#garph b\w modle and r2
plt.figure(figsize= (10 , 6))

sns.barplot(x='Models', y='R2-Score', data = result , palette = 'muted')

plt.xticks(rotation=45)
plt.title("Model vs R2 Score")
plt.xlabel("Models")
plt.ylabel("R2 Score")

plt.show()
```

```python
#plotting all error on model
plt.figure(figsize = (18 , 6))
def plotting(var , num):
    plt.subplot(1 , 3 , num)
    sns.barplot(x = 'Models' , y = var , data = result , palette= 'inferno')
    plt.title(var)
    plt.xticks(rotation = 45)

plotting('MAE-Error' , 1)
plotting('MSE-Error' , 2)
plotting('RMSE-Error' , 3)

plt.tight_layout()
plt.show()
```

```python
combined = result.melt(
    id_vars="Models",
    value_vars=["R2-Score", "MAE-Error",  "RMSE-Error"],
    var_name="Metric",
    value_name="Value"
)

plt.figure(figsize=(14, 6))

ax = sns.barplot(
    x="Models",
    y="Value",
    hue="Metric",
    data=combined,
    palette="Set2"
)

plt.title("Model Comparison (R2 vs MAE vs RMSE)")
plt.xticks(rotation=45)
plt.xlabel("Models")
plt.ylabel("Score / Error")

for i in ax.containers:
    ax.bar_label(i, fmt="%.2f")

plt.legend(title="Metrics")
plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(18, 10))

def plottingg(pred, title, num):
    plt.subplot(2, 3, num)
    plt.scatter(y_test, pred)
    plt.plot([y_test.min(), y_test.max()],
             [y_test.min(), y_test.max()],
             color='red') 
    plt.title(title)
    plt.xlabel("Actual")
    plt.ylabel("Predicted")


plottingg(L_pred, "Linear Regression", 1)
plottingg(KN_pred, "KNN", 2)
plottingg(T_pred, "Decision Tree", 3)
plottingg(R_pred, "Random Forest", 4)
plottingg(S_pred, "SVM", 5)
plottingg(RID_pred, "Ridge", 6)

plt.tight_layout()
plt.show()
```
