## Outliers And SkewNess

```python
#outliers
numeric_cols = [
    'price',
    'minimum_nights',
    'number_of_reviews',
    'reviews_per_month',
    'calculated_host_listings_count',
    'availability_365',
    'latitude',
    'longitude'
]
data[numeric_cols].describe().T
```

```python
num_featture = data.select_dtypes(include = 'number')

Q1 = num_featture.quantile(0.25)
Q3 = num_featture.quantile(0.75)
        
IQR = Q3 - Q1
Lower_Bond = Q1 - 1.5 * IQR
Upper_Bond = Q3 + 1.5 * IQR

outliers = (num_featture < Lower_Bond) | (num_featture > Upper_Bond)
print(outliers.sum())
```

```python
data.columns
```

```python
# HAnding the Outliers 
# in price', 'minimum_nights', 'number_of_reviews', 'reviews_per_month', 'calculated_host_listings_count'
# left 'latitude', 'longitude', cuz there are good in (-ve) and (+ve) values

#using np.log1p

data = data[data['price'] > 0].copy()
data['price_log'] = np.log1p(data['price'])

data['minimum_nights_log'] = np.log1p(data['minimum_nights'])

data['number_of_reviews_log'] = np.log1p(data['number_of_reviews'])

data['reviews_per_month_log'] = np.log1p(data['reviews_per_month'])

data['calculated_host_listings_count_log'] = np.log1p(data['calculated_host_listings_count'])
```

```python
# analysic the orignal and log columns with shew values 
original_features = [
    'price',
    'minimum_nights',
    'number_of_reviews',
    'reviews_per_month',
    'calculated_host_listings_count'
]

log_features = [
    'price_log',
    'minimum_nights_log',
    'number_of_reviews_log',
    'reviews_per_month_log',
    'calculated_host_listings_count_log'
]

print("++++++++BEFORE TRANSFORMATION++++++++")
print(data[original_features].skew())

print("\n++++++++AFTER TRANSFORMATION+++++++++++")
print(data[log_features].skew())
```

```python
data.columns
```

```python
numerical_features = [                            
    'latitude',               # original
    'longitude',              #  original
    'price',                  # original
    'minimum_nights_log',     # transformed
    'number_of_reviews',      # original
    'reviews_per_month_log',   # transformed
    'calculated_host_listings_count',  # original
    'availability_365'                  # original
]

categorical_features = [
    'neighbourhood_group', # original
    'neighbourhood'        # original
]
```

```python
print("Duplicate rows:", data.duplicated().sum())
```

## TRAIN AND TEST


```python
# train test split and doing value on x and y
from sklearn.model_selection import train_test_split

x = data[numerical_features + categorical_features]
y = data['room_type']

x_train, x_test , y_train , y_test = train_test_split(x , y , test_size= 0.2 , random_state= 42 , shuffle= True , stratify=y)
```

```python
# cheking shapes of traing and testing values 
print("X shape:", x.shape)

print("\nX_train:", x_train.shape)
print("X_test :", x_test.shape)

print("\ny_train:", y_train.shape)
print("y_test :", y_test.shape)
```

## PipeLine and ColumnTransformer

```python
from sklearn.preprocessing import StandardScaler , OneHotEncoder 
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer  #  --->  this work when any null value are predent in data set then its fill UP
from sklearn.compose import ColumnTransformer

numerical_pipeline = Pipeline(steps = [
    ('impute' ,  SimpleImputer(strategy= 'median')),
    ('scale' , StandardScaler())
])

categorical_pipeline = Pipeline(steps = [
    ('impute' , SimpleImputer(strategy = 'most_frequent')),
    ('scale' , OneHotEncoder(handle_unknown= 'ignore'))
])

preprocessor = ColumnTransformer(transformers=[
    ('numerical' , numerical_pipeline , numerical_features),
    ('categorical' , categorical_pipeline , categorical_features)
])
preprocessor
```

## Model Traning

```python
# Logistic-Regressor
from sklearn.linear_model import LogisticRegression
LR_pipeline = Pipeline(steps = [
    ('preprocessing' , preprocessor),
    ('model' , LogisticRegression(
        class_weight="balanced",
        random_state=42,
        max_iter=1000,
    ))
])

LR_pipeline.fit(x_train , y_train)
LR_predict = LR_pipeline.predict(x_test)
print("Logistic Prediction : " , LR_predict[:10])
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score,confusion_matrix,classification_report
# Logistic Regression Metrics

LR_Accuracy = accuracy_score(y_test,LR_predict)

LR_Precision = precision_score(y_test,LR_predict,average='macro')

LR_Recall = recall_score(y_test,LR_predict,average='macro')

LR_F1 = f1_score(y_test,LR_predict,average='macro')

LR_Confusion_Matrix = confusion_matrix(y_test,LR_predict)

print("Logistic Regression Accuracy :", LR_Accuracy)
print("Logistic Regression Precision:", LR_Precision)
print("Logistic Regression Recall   :", LR_Recall)
print("Logistic Regression Macro F1 :", LR_F1)

print("\nClassification Report:")
print(classification_report(y_test,LR_predict))

print("\nConfusion Matrix:")
print(LR_Confusion_Matrix)
```

```python
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(y_test, LR_predict, xticks_rotation=45)

plt.title("Logistic Regression - Confusion Matrix")
plt.show()
```

```python
# Descision Tree
from sklearn.tree import DecisionTreeClassifier
Tree_pipeline = Pipeline(steps= [
    ('preprocessing' , preprocessor),
    ('model' , DecisionTreeClassifier(
        criterion='gini',
        max_depth=10,
        min_samples_split=10,
        min_samples_leaf=5,
        class_weight='balanced',
        random_state=42 
    ))
])

Tree_pipeline.fit(x_train , y_train)
Tree_predict = Tree_pipeline.predict(x_test)

print("Tree Prediction Value : " , Tree_predict[:10])
```

```python
# Decision Tree  Metrics

Tree_Accuracy = accuracy_score(y_test,Tree_predict)

Tree_Precision = precision_score(y_test,Tree_predict,average='macro')

Tree_Recall = recall_score(y_test,Tree_predict,average='macro')

Tree_F1 = f1_score(y_test,Tree_predict,average='macro')

Tree_Confusion_Matrix = confusion_matrix(y_test,Tree_predict)

print("Logistic Regression Accuracy :", Tree_Accuracy)
print("Logistic Regression Precision:", Tree_Precision)
print("Logistic Regression Recall   :", Tree_Recall)
print("Logistic Regression Macro F1 :", Tree_F1)

print("\nClassification Report:")
print(classification_report(y_test,Tree_predict))

print("\nConfusion Matrix:")
print(Tree_Confusion_Matrix)
```

```python
#display the Confusion Metrix OF TREE
from sklearn.metrics import ConfusionMatrixDisplay
plt.figure(figsize= (8 , 6))
ConfusionMatrixDisplay.from_predictions(y_test , Tree_predict , xticks_rotation= 45)

plt.title("Decision Tree - Confusion Matrix")
plt.show()
```

```python
# Random Forest Classifiers
from sklearn.ensemble import RandomForestClassifier
RF_pipeline = Pipeline(steps = [
    ('preprocessing' , preprocessor),
    ('model' , RandomForestClassifier(
        n_estimators=200,
        criterion='gini',
        max_depth=15,
        min_samples_split=10,
        min_samples_leaf=5,
        max_features='sqrt',
        class_weight='balanced',
        random_state=42,
        n_jobs=-1
    ))
])

RF_pipeline.fit(x_train , y_train)
RF_predict = RF_pipeline.predict(x_test)
print("Prediction VAlue : " , RF_predict[:10])
```

```python
# Decision RF  Metrics

RF_Accuracy = accuracy_score(y_test,RF_predict)

RF_Precision = precision_score(y_test,RF_predict,average='macro')

RF_Recall = recall_score(y_test,RF_predict,average='macro')

RF_F1 = f1_score(y_test,RF_predict,average='macro')

RF_Confusion_Matrix = confusion_matrix(y_test,RF_predict)

print("Logistic Regression Accuracy :", RF_Accuracy)
print("Logistic Regression Precision:", RF_Precision)
print("Logistic Regression Recall   :", RF_Recall)
print("Logistic Regression Macro F1 :", RF_F1)

print("\nClassification Report:")
print(classification_report(y_test,RF_predict))

print("\nConfusion Matrix:")
print(RF_Confusion_Matrix)
```

```python
#display the Confusion Metrix OF Forest
from sklearn.metrics import ConfusionMatrixDisplay
plt.figure(figsize= (8 , 6))
ConfusionMatrixDisplay.from_predictions(y_test , RF_predict , xticks_rotation= 45)

plt.title("Random Forest - Confusion Matrix")
plt.show()
```

## When i use to XgBoost Model train its requid Leable Encoding oF Y value and and aing train test then i Predict the value 

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(data['room_type'])
print(label_encoder.classes_)
```

```python
x = data[numerical_features + categorical_features]
y = y_encoded

X_train, X_test, y_train, y_test = train_test_split(x,y,test_size=0.20,random_state=42,shuffle=True,stratify=y)
# X is capital and Also Y cus clarifying which is Previous and which is next (small letter is previous and captal is new train modle with y is LB enocding)
```

```python
from xgboost import XGBClassifier
from sklearn.pipeline import Pipeline

XGB_pipeline = Pipeline(steps=[
    ('preprocessing', preprocessor),
    ('model', XGBClassifier(
        n_estimators=200,
        learning_rate=0.05,
        max_depth=6,
    ))
])

XGB_pipeline.fit(X_train,y_train)
XGB_predict = XGB_pipeline.predict(X_test)
```

```python
# XGBoost Metrics

XGB_Accuracy = accuracy_score(y_test,XGB_predict)
XGB_Precision = precision_score(y_test,XGB_predict,average='macro')
XGB_Recall = recall_score(y_test,XGB_predict,average='macro')

XGB_F1 = f1_score(y_test,XGB_predict,average='macro')

XGB_Confusion_Matrix = confusion_matrix(y_test,XGB_predict)

print("XGBoost Accuracy :", XGB_Accuracy)
print("XGBoost Precision:", XGB_Precision)
print("XGBoost Recall   :", XGB_Recall)
print("XGBoost Macro F1 :", XGB_F1)

print("\nClassification Report:")
print(classification_report(y_test,XGB_predict,target_names=label_encoder.classes_))

print("\nConfusion Matrix:")
print(XGB_Confusion_Matrix)
```

```python
#display the Confusion Metrix OF Forest
from sklearn.metrics import ConfusionMatrixDisplay
plt.figure(figsize= (8 , 6))
ConfusionMatrixDisplay.from_predictions(y_test , XGB_predict , xticks_rotation= 45)

plt.title("XG-BOOST - Confusion Matrix")
plt.show()
```

```python
# Comparing
model_comparison = pd.DataFrame({
    'Model': ['Logistic Regression','Decision Tree','Random Forest','XGBoost'],
    
    'Accuracy': [LR_Accuracy,Tree_Accuracy,RF_Accuracy,XGB_Accuracy],

    'Precision': [LR_Precision,Tree_Precision,RF_Precision,XGB_Precision],

    'Recall': [LR_Recall,Tree_Recall,RF_Recall,XGB_Recall],

    'Macro F1': [LR_F1,Tree_F1,RF_F1,XGB_F1]
})

# Sort by Macro F1
model_comparison = model_comparison.sort_values(by='Macro F1',ascending=False).reset_index(drop=True)
print(model_comparison)
```

```python
# Analysic the traning and testing parts
print("Training class distribution:")

class_counts = np.bincount(y_train)

for i, count in enumerate(class_counts):
    print(f"{i} ({label_encoder.classes_[i]}): {count}")

total = len(y_train)

print("\nTraining class percentage:")

for i, count in enumerate(class_counts):
    percentage = (count / total) * 100

    print(f"{i} ({label_encoder.classes_[i]}): "f"{count} ({percentage:.2f}%)")    
```

```python
# SMOTE apply and train agian on XGboot model
from imblearn.pipeline import Pipeline
from imblearn.over_sampling import SMOTE
XGB_SMOTE_pipeline = Pipeline(steps=[
    ('preprocessing', preprocessor),
    ('smote', SMOTE(random_state=42)),
    ('model', XGBClassifier(
        n_estimators=200,
        learning_rate=0.05,
        max_depth=6,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42,
        eval_metric='mlogloss'
    ))
])

XGB_SMOTE_pipeline.fit(X_train,y_train)
XGB_SMOTE_predict = XGB_SMOTE_pipeline.predict(X_test)
```

```python
# Score Of xGBoost with Smote Apply 
XGB_SMOTE_Accuracy = accuracy_score(y_test,XGB_SMOTE_predict)

XGB_SMOTE_Precision = precision_score(y_test,XGB_SMOTE_predict,average='macro')

XGB_SMOTE_Recall = recall_score(y_test,XGB_SMOTE_predict,average='macro')

XGB_SMOTE_F1 = f1_score(y_test,XGB_SMOTE_predict,average='macro')

print("XGBoost + SMOTE Accuracy :", XGB_SMOTE_Accuracy)
print("XGBoost + SMOTE Precision:", XGB_SMOTE_Precision)
print("XGBoost + SMOTE Recall   :", XGB_SMOTE_Recall)
print("XGBoost + SMOTE Macro F1 :", XGB_SMOTE_F1)

print("\nClassification Report:")
print(classification_report(y_test,XGB_SMOTE_predict,target_names=label_encoder.classes_))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test,XGB_SMOTE_predict))
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
plt.figure(figsize= (8 , 6))
ConfusionMatrixDisplay.from_predictions(y_test ,XGB_SMOTE_predict , xticks_rotation= 45)

plt.title("XGB_SMOTE_predict - Confusion Matrix")
plt.show()
```

```python
import pandas as pd

xgb_smote_comparison = pd.DataFrame({'Model': ['XGBoost','XGBoost + SMOTE'],

    'Accuracy': [XGB_Accuracy,XGB_SMOTE_Accuracy],
    'Precision': [XGB_Precision,XGB_SMOTE_Precision],
    'Recall': [XGB_Recall,XGB_SMOTE_Recall],
    'Macro F1': [XGB_F1,XGB_SMOTE_F1]
})

print(xgb_smote_comparison)
```

## Hyperparameter Tuning

```python
from sklearn.model_selection import RandomizedSearchCV
param_distributions = {
    'model__n_estimators': [100, 200, 300, 400],
    'model__learning_rate': [0.01, 0.03, 0.05, 0.1],
    'model__max_depth': [3, 5, 6, 7, 9],
    'model__min_child_weight': [1, 3, 5],
    'model__subsample': [0.7, 0.8, 0.9, 1.0],
    'model__colsample_bytree': [0.7, 0.8, 0.9, 1.0],
    'model__gamma': [0, 0.1, 0.3, 0.5]
}

random_search = RandomizedSearchCV(
    estimator=XGB_pipeline,
    param_distributions=param_distributions,
    n_iter=20,
    scoring='f1_macro',
    cv=3,
    random_state=42,
    n_jobs=-1,
    verbose=1
)
```

```python
random_search.fit(X_train,y_train)
```

```python
print("Best Parameters:")
print(random_search.best_params_)

print("Best CV Macro F1:")
print(random_search.best_score_)
```

```python
# Predict the best values 
best_xgb = random_search.best_estimator_
XGB_Tuned_predict = best_xgb.predict(X_test)

# Metics the best Parameter of XGboost
XGB_Tuned_Accuracy = accuracy_score(y_test,XGB_Tuned_predict)
XGB_Tuned_Precision = precision_score(y_test,XGB_Tuned_predict,average='macro')
XGB_Tuned_Recall = recall_score(y_test,XGB_Tuned_predict,average='macro')

XGB_Tuned_F1 = f1_score(y_test,XGB_Tuned_predict,average='macro')

print("Tuned XGBoost Accuracy :", XGB_Tuned_Accuracy)
print("Tuned XGBoost Precision:", XGB_Tuned_Precision)
print("Tuned XGBoost Recall   :", XGB_Tuned_Recall)
print("Tuned XGBoost Macro F1 :", XGB_Tuned_F1)

print("Final Tuned XGBoost Classification Report:")

print(classification_report(y_test,XGB_Tuned_predict,))

print("Final Confusion Matrix:")

print(confusion_matrix(y_test,XGB_Tuned_predict))
```

```python
plt.figure(figsize= (8 , 6))
ConfusionMatrixDisplay.from_predictions(y_test ,XGB_Tuned_predict , xticks_rotation= 45)

plt.title("XGB_Tuned_predict - Confusion Matrix")
plt.show()
```
